```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


class SalesAnalyzer:

    def __init__(self):
        self.data = None
        print("==== Data Analysis & Visualization Program ====")


    def loadData(self):

        print("== Load Dataset ==")

        try:
            path = input("Enter CSV File Path: ")
            self.data = pd.read_csv(path)
            print("Dataset Loaded Successfully!")

        except:
            print("Error: File Not Found!")



    def exploreData(self):

        if self.data is None:
            print("Please Load Dataset First!")
            return


        print("== Explore Data ==")
        print("1. Display First 5 Rows")
        print("2. Display Last 5 Rows")
        print("3. Display Column Names")
        print("4. Display Data Types")
        print("5. Display Info")
        print("6. Back")


        while True:

            ch = int(input("Enter Choice: "))


            if ch == 1:
                print(self.data.head())

            elif ch == 2:
                print(self.data.tail())

            elif ch == 3:
                print(self.data.columns)

            elif ch == 4:
                print(self.data.dtypes)

            elif ch == 5:
                print(self.data.info())

            elif ch == 6:
                break

            else:
                print("Invalid Choice!")

    def dataOperations(self):

        if self.data is None:
            print("Please Load Dataset First!")
            return


        print("=== DataFrame Operations ===")
        print("1. Sort by Total Amount")
        print("2. Filter Amount Greater Than Value")
        print("3. Show Top 5 Sales")
        print("4. Show Average Sales")
        print("5. Back")


        while True:

            ch = int(input("Enter Choice: "))

            if ch == 1:

                result = self.data.sort_values("Total_Amount", ascending=False)

                print("Sorted Data:\n", result)


            
            elif ch == 2:

                val = float(input("Enter Amount: "))

                print(self.data[self.data["Total_Amount"] > val])


            
            elif ch == 3:

                print("Top 5 Sales Records:")
                print(self.data.nlargest(5, "Total_Amount"))


           
            elif ch == 4:

                avg = self.data["Total_Amount"].mean()

                print("Average Sales Amount =", avg)


            elif ch == 5:
                break


            else:
                print("Invalid Choice!")



    def handleMissing(self):

        if self.data is None:
            print("Please Load Dataset First!")
            return


        print("=== Handle Missing Data ===")
        print("1. Show Missing Rows")
        print("2. Fill with Mean")
        print("3. Delete Missing Rows")
        print("4. Replace with Value")
        print("5. Back")


        while True:

            ch = int(input("Enter Choice: "))


            if ch == 1:

                print(self.data[self.data.isnull().any(axis=1)])


            elif ch == 2:

                self.data.fillna(self.data.mean(numeric_only=True), inplace=True)

                print("Missing Values Filled!")


            elif ch == 3:

                self.data.dropna(inplace=True)

                print("Rows Deleted!")


            elif ch == 4:

                val = input("Enter Value: ")

                self.data.fillna(val, inplace=True)

                print("Values Replaced!")


            elif ch == 5:

                break



    def descriptive(self):

        if self.data is None:
            print("Please Load Dataset First!")
            return


        print("=== Descriptive Statistics ===")

        print(self.data.describe())



    def visualization(self):

        if self.data is None:
            print("Please Load Dataset First!")
            return


        self.data["Date"] = pd.to_datetime(self.data["Date"], errors="coerce")

        self.data = self.data.sort_values("Date")


        print("==== Data Visualization ====")
        print("1. Bar Plot")
        print("2. Line Plot")
        print("3. Scatter Plot")
        print("4. Pie Plot")
        print("5. Histogram Plot")

        Ent_v = int(input("Enter Your Choice: "))


        if Ent_v == 1:

            city_sales = self.data.groupby("City")["Total_Amount"].sum()

            plt.figure(figsize=(10,5))
            plt.bar(city_sales.index, city_sales.values)

            plt.title("Total Sales by City")
            plt.xlabel("City")
            plt.ylabel("Total Amount")

            plt.xticks(rotation=45)
            plt.show()



        elif Ent_v == 2:

            date_sales = self.data.groupby("Date")["Total_Amount"].sum()

            plt.figure(figsize=(12,6))
            plt.plot(date_sales.index, date_sales.values, marker="o")

            plt.title("Sales Trend (Line Chart)")
            plt.xlabel("Date")
            plt.ylabel("Total Amount")

            plt.xticks(rotation=45)
            plt.grid(True)
            plt.show()

        elif Ent_v == 3:

            plt.figure(figsize=(8,5))
            plt.scatter(self.data["Quantity"],
                        self.data["Total_Amount"])

            plt.title("Quantity vs Total Amount")
            plt.xlabel("Quantity")
            plt.ylabel("Total Amount")

            plt.show()

        elif Ent_v == 4:

            product_sales = self.data.groupby("Product")["Total_Amount"].sum()

            plt.figure(figsize=(7,7))
            plt.pie(product_sales,
                    labels=product_sales.index,
                    autopct="%1.1f%%",
                    startangle=90)

            plt.title("Product Wise Sales")
            plt.show()

        elif Ent_v == 5:

            plt.figure(figsize=(8,5))
            plt.hist(self.data["Total_Amount"], bins=10)

            plt.title("Sales Distribution")
            plt.xlabel("Total Amount")
            plt.ylabel("Frequency")

            plt.show()

        else:
            print("Invalid Choice!")

    def savePlot(self):

        name = input("Enter File Name (ex: plot.png): ")

        try:

            plt.savefig(name)
            print("Plot Saved Successfully!")

        except:

            print("Error Saving File!")

def main():

    obj = SalesAnalyzer()


    while True:

        print("\n===== MAIN MENU =====")
        print("1. Load Dataset")
        print("2. Explore Data")
        print("3. Perform Operation")
        print("4. Handle Missing Data")
        print("5. Descriptive Statistics")
        print("6. Data Visualization")
        print("7. Save Plot")
        print("8. Exit")

        ch = int(input("Enter Choice: "))

        if ch == 1:
            obj.loadData()

        elif ch == 2:
            obj.exploreData()

        elif ch == 3:
            obj.dataOperations()     

        elif ch == 4:
            obj.handleMissing()

        elif ch == 5:
            obj.descriptive()

        elif ch == 6:
            obj.visualization()

        elif ch == 7:
            obj.savePlot()

        elif ch == 8:
            print("Program Closed!")
            break

        else:
            print("Invalid Choice!")

main()
```
