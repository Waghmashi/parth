````python
print("--- Part C: Data Understanding & Cleaning ---")

print("\n--- Initial Info ---")
print(master_df.info())
print("\n--- Describe ---")
print(master_df.describe())

master_df.loc[0:5, 'Income'] = np.nan
master_df.loc[10, 'Age'] = np.nan
master_df = pd.concat([master_df, master_df.iloc[0:2]], ignore_index=True) # Add duplicates

print(f"\nMissing values before cleaning:\n{master_df.isnull().sum()}")
print(f"Duplicates before cleaning: {master_df.duplicated().sum()}")

master_df['Income'] = master_df['Income'].fillna(master_df['Income'].median())
master_df['Age'] = master_df['Age'].fillna(master_df['Age'].mean())

master_df.drop_duplicates(inplace=True)

master_df['Churn'] = master_df['Churn'].astype('category')

print(f"\nMissing values after cleaning:\n{master_df.isnull().sum()}")
print(f"Duplicates after cleaning: {master_df.duplicated().sum()}")
````
